# EXP001 — Baseline Source Separation using HT-Demucs

## Experiment Overview

This notebook implements **EXP001**, the first controlled experiment in the IKS Internship project titled **"Multitrack Source Separation and Instrument Isolation for Indian Folk Music."**

The objective of this experiment is to establish a **baseline performance** of the pre-trained **HT-Demucs** source separation model without any fine-tuning or domain adaptation. The outputs generated from this experiment serve as reference results for evaluating future models and processing pipelines.

The experiment processes one selected recording from the **BIHU** folk music dataset and separates the mixture into its constituent audio stems produced by HT-Demucs.

# 9. Source Separation

This section performs vocal and accompaniment separation using the selected HT-Demucs model.

For each selected recording, the notebook:

- Executes HT-Demucs source separation.
- Generates the separated stems.
- Extracts the vocals and accompaniment.
- Renames the outputs according to the project naming convention.
- Stores the separated audio files in the experiment output directory.

The separated accompaniment (`No_Vocals`) will be used in future experiments for instrument isolation and analysis.

````python
# ============================================================================
# SOURCE SEPARATION
# ============================================================================
# Run HT-Demucs on the selected recordings.
# ============================================================================

import subprocess

print("=" * 70)
print(f"Running HT-Demucs ({MODEL_NAME})")
print("=" * 70)

for index, recording in enumerate(SELECTED_RECORDINGS, start=1):

    print(f"\n[{index}/{len(SELECTED_RECORDINGS)}] Processing: {recording.name}")

    command = [
        "python",
        "-m",
        "demucs",
        "--two-stems",
        "vocals",
        "-n",
        MODEL_NAME,
        "-o",
        str(OUTPUT_ROOT),
        str(recording)
    ]

    subprocess.run(command, check=True)

print("\n" + "=" * 70)
print("HT-Demucs source separation completed successfully.")
print("=" * 70)
````

# 10. Output Organization

This section reorganizes the raw HT-Demucs outputs into the standardized project structure.

The process includes:

- Locating the separated stems generated by HT-Demucs.
- Extracting the vocal and accompaniment files.
- Renaming the files according to the project naming convention.
- Moving the files into the appropriate experiment output directories.
- Removing temporary files created by HT-Demucs.

After this step, the experiment outputs are organized consistently and are ready for analysis or subsequent processing.

````python
# ============================================================================
# OUTPUT ORGANIZATION
# ============================================================================
# Rename and organize the HT-Demucs outputs into the project structure.
# ============================================================================

import shutil

print("=" * 70)
print("Organizing Outputs")
print("=" * 70)

for recording in SELECTED_RECORDINGS:

    recording_id = recording.stem

    vocals_source = next(
        OUTPUT_ROOT.glob(f"**/{recording_id}/vocals.wav"),
        None
    )

    accompaniment_source = next(
        OUTPUT_ROOT.glob(f"**/{recording_id}/no_vocals.wav"),
        None
    )

    if vocals_source is None:
        raise FileNotFoundError(
            f"Could not locate vocals.wav for {recording.name}"
        )

    if accompaniment_source is None:
        raise FileNotFoundError(
            f"Could not locate no_vocals.wav for {recording.name}"
        )

    vocals_destination = (
        VOCALS_OUTPUT_DIR /
        f"{recording_id}{VOCAL_SUFFIX}"
    )

    accompaniment_destination = (
        NO_VOCALS_OUTPUT_DIR /
        f"{recording_id}{NO_VOCAL_SUFFIX}"
    )

    shutil.move(str(vocals_source), str(vocals_destination))
    shutil.move(str(accompaniment_source), str(accompaniment_destination))

    print(f"✓ {recording.name}")

# Remove temporary Demucs output folders
for model_directory in OUTPUT_ROOT.iterdir():

    if (
        model_directory.is_dir()
        and model_directory.name == MODEL_NAME
    ):
        shutil.rmtree(model_directory)

print("\n" + "=" * 70)
print("Outputs organized successfully.")
print("=" * 70)
````
